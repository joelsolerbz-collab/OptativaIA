````python
import tensorflow as tf
from keras.models import load_model
from PIL import Image, ImageOps
import numpy as np

class_names = open("labels.txt", "r").readlines()
data = np.ndarray(shape=(1, 224, 224, 3), dtype=np.float32)
image = Image.open("incidente.png").convert("RGB")
size = (224, 224)
image = ImageOps.fit(image, size, Image.Resampling.LANCZOS)
image_array = np.asarray(image)
normalized_image_array = (image_array.astype(np.float32) / 127.5) - 1
data[0] = normalized_image_array

prediction = model.predict(data)
index = np.argmax(prediction)
class_name = class_names[index]
confidence_score = prediction[0][index]

# Print prediction and confidence score
print("Class:", class_name[2:], end="")
print("Confidence Score:", confidence_score)
````

````python
# implementa la clasificación con la imagen que pertenece claramente a una categoría
from PIL import Image, ImageOps
import numpy as np

class_names = open("labels.txt", "r").readlines()

def clasificar(ruta):
    data = np.ndarray(shape=(1, 224, 224, 3), dtype=np.float32)
    image = Image.open(ruta).convert("RGB")
    image = ImageOps.fit(image, (224, 224), Image.Resampling.LANCZOS)
    data[0] = (np.asarray(image).astype(np.float32) / 127.5) - 1

    prediction = model.predict(data)
    index = np.argmax(prediction)
    print("Imagen:", ruta)
    print("Clase:", class_names[index][2:].strip())
    print("Confianza:", round(float(prediction[0][index]) * 100, 2), "%")
    print("-" * 30)

clasificar("clara.jpg")
````

````python
# implementa la clasificación con una imagen ambigua, que puede ser de cualquier categoría
clasificar("ambigua.jpg")
````

````python
# implementa la clasificación con una imagen sin ninguna categoría del modelo
clasificar("nada.jpg")
````

**¿Cómo se usa el modelo exportado? ¿Cómo se comporta?**

---



**¿Cómo se usa el modelo exportado?**

Primero se exporta el modelo desde Teachable Machine para TensorFlow y Keras, y eso te da dos archivos: keras_model.h5, que es el modelo ya entrenado, y labels.txt, que tiene los nombres de las clases. Luego en Colab se carga el modelo, se lee la lista de clases y a cada imagen se le hace lo mismo: se pasa a RGB, se recorta y se ajusta a 224x224 y se normaliza. Después se la damos al modelo con predict y nos devuelve una probabilidad por cada clase. La más alta es la que elige como respuesta, y ese valor es la confianza



---


**¿Cómo se comporta?**


Con la imagen clara clara predijo cactus con un 86,37 % de confianza, así que acertó y estaba bastante seguro



Con la imagen ambigua ambigua predijo mechero con un 78,54 %. La confianza bajó un poco, lo que indica que dudaba más


Con la imagen que no pertenece a ninguna clase nada predijo mechero con un 96,14 %. Lo curioso es que esta fue la confianza más alta, aun siendo una imagen que el modelo no conoce


Esto pasa porque el modelo solo sabe las clases con las que lo entrené y siempre tiene que repartir el 100 % entre ellas. No tiene la opción de decir "esto no es ninguna de mis clases". Por eso una confianza alta no siempre significa que la respuesta sea correcta. Para mejorarlo se podría añadir una clase "otros" con fotos variadas, entrenar con más imágenes distintas o poner un mínimo de confianza para aceptar una predicción
